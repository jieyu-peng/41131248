````python
!pip install -q groq

import os
from google.colab import userdata
from groq import Groq
import sqlite3

# 1. 初始化 Groq 客戶端
client = Groq(api_key=userdata.get('GROQ_API_KEY'))

# 2. 建立 SQLite 資料庫與初始資料
def init_db():
    conn = sqlite3.connect('activities.db')
    cursor = conn.cursor()

    # 建立活動表
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS activities (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT UNIQUE,
            date TEXT,
            capacity INTEGER
        )
    ''')

    # 建立報名表
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS registrations (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT,
            activity_name TEXT,
            count INTEGER,
            status TEXT
        )
    ''')

    # 寫入初始活動資料（若已存在則忽略）
    activities_data = [
        ('迎新晚會', '2026-11-20', 6),
        ('程式工作坊', '2026-11-25', 4),
        ('籃球友誼賽', '2026-11-28', 3)
    ]
    cursor.executemany('''
        INSERT OR IGNORE INTO activities (name, date, capacity)
        VALUES (?, ?, ?)
    ''', activities_data)

    conn.commit()
    conn.close()

init_db()
print("資料庫初始化完成！")
````

````python
def process_request(student_name, user_input):
    # 1. 透過 Groq AI 解析語意（使用第一格定義的 MODEL 變數）
    prompt = f"""
    你是一個系學會活動助理。請解析使用者的一句話，嚴格以「動作,活動,人數」格式輸出，不要有多餘字眼。
    - 動作只能是：報名、取消、查詢、統計、其他
    - 活動只能是：迎新晚會、程式工作坊、籃球友誼賽、無
    - 人數：預設為 1，若提及多人請萃取對應數字（1～3）
    - 若無法判斷，輸出：其他,無,1

    使用者說：「{user_input}」
    """

    response = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=0
    )

    res_text = response.choices[0].message.content.strip()
    parts = res_text.split(',')
    if len(parts) < 3:
        return "AI 解析格式錯誤", res_text

    action, activity, count_str = parts[0].strip(), parts[1].strip(), parts[2].strip()
    try:
        count = int(count_str)
    except:
        count = 1

    conn = sqlite3.connect('activities.db')
    cursor = conn.cursor()

    # 2. 根據動作執行資料庫邏輯與規則檢查
    if action == "報名":
        if not (1 <= count <= 3):
            conn.close()
            return f"失敗：每次限報 1～3 人（您輸入了 {count} 人）", res_text

        cursor.execute("SELECT capacity FROM activities WHERE name = ?", (activity,))
        act_row = cursor.fetchone()
        if not act_row:
            conn.close()
            return f"失敗：找不到活動「{activity}」", res_text
        capacity = act_row[0]

        cursor.execute("SELECT * FROM registrations WHERE student_name = ? AND activity_name = ? AND status = '已報名'", (student_name, activity))
        if cursor.fetchone():
            conn.close()
            return f"失敗：您已經報名過「{activity}」了", res_text

        cursor.execute("SELECT COUNT(DISTINCT activity_name) FROM registrations WHERE student_name = ? AND status = '已報名'", (student_name,))
        active_count = cursor.fetchone()[0]
        if active_count >= 2:
            conn.close()
            return f"失敗：每人最多同時報名 2 個活動", res_text

        cursor.execute("SELECT SUM(count) FROM registrations WHERE activity_name = ? AND status = '已報名'", (activity,))
        booked_sum = cursor.fetchone()[0] or 0
        left_spots = capacity - booked_sum
        if count > left_spots:
            conn.close()
            return f"失敗：名額不足（剩餘名額：{left_spots} 人）", res_text

        cursor.execute("INSERT INTO registrations (student_name, activity_name, count, status) VALUES (?, ?, ?, '已報名')",
                       (student_name, activity, count))
        conn.commit()
        conn.close()
        return f"報名成功！「{activity}」共 {count} 人", res_text

    elif action == "取消":
        cursor.execute("SELECT * FROM registrations WHERE student_name = ? AND activity_name = ? AND status = '已報名'", (student_name, activity))
        reg = cursor.fetchone()
        if not reg:
            conn.close()
            return f"失敗：查無您報名「{activity}」的有效紀錄", res_text

        cursor.execute("UPDATE registrations SET status = '已取消' WHERE student_name = ? AND activity_name = ? AND status = '已報名'", (student_name, activity))
        conn.commit()
        conn.close()
        return f"取消成功！已釋出「{activity}」的名額", res_text

    elif action == "查詢":
        cursor.execute("SELECT capacity FROM activities WHERE name = ?", (activity,))
        act_row = cursor.fetchone()
        if not act_row:
            conn.close()
            return f"失敗：找不到活動「{activity}」", res_text
        capacity = act_row[0]

        cursor.execute("SELECT SUM(count) FROM registrations WHERE activity_name = ? AND status = '已報名'", (activity,))
        booked_sum = cursor.fetchone()[0] or 0
        left_spots = capacity - booked_sum
        conn.close()
        return f"查詢結果：{activity} 總容量 {capacity} 人，目前剩餘 {left_spots} 個名額", res_text

    elif action == "統計":
        cursor.execute("SELECT name, capacity FROM activities")
        acts = cursor.fetchall()
        report = "【各活動報名統計】\n"
        for name, cap in acts:
            cursor.execute("SELECT SUM(count) FROM registrations WHERE activity_name = ? AND status = '已報名'", (name,))
            booked = cursor.fetchone()[0] or 0
            report += f"- {name}: 已報名 {booked} 人 / 上限 {cap} 人\n"
        conn.close()
        return report.strip(), res_text

    else:
        conn.close()
        return "對不起，我聽不太懂您的意思，請重新輸入有關報名、查詢或取消的指令。", res_text

print("核心函數已成功對接 MODEL 變數！")
````

````python
def chatbot_interface(student_name, user_input):
    # 呼叫你寫好的核心邏輯函式
    response_msg, raw_text = process_request(student_name, user_input)
    return response_msg
````

````python
import gradio as gr

# 建立 Gradio 介面
demo = gr.Interface(
    fn=chatbot_interface,                              # 綁定的後端函數
    inputs=[
        gr.Textbox(label="學生姓名", value="小明"),          # 輸入欄位 1：姓名
        gr.Textbox(label="請輸入您的需求（例如：我要報名迎新晚會）", lines=2) # 輸入欄位 2：指令
    ],
    outputs=gr.Textbox(label="系統回應"),                 # 輸出欄位：顯示系統結果
    title="系學會活動報名系統",                            # 網頁標題
    description="請輸入您的姓名與需求，系統將自動為您處理活動報名、取消或查詢。" # 網頁說明
)

# 啟動網頁 (在 Colab 執行時會直接產生互動視窗)
demo.launch()
````
